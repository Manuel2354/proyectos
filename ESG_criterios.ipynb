```python
# Top 5 de México
top_mexico = esg_completo[esg_completo['Mercado '] == 'MEX'].sort_values(
    by=['S&P Global ESG Score', 'Rendimiento'], ascending=False).head(5)
```

```python
# Top 5 de EUA
top_EUA = esg_completo[esg_completo['Mercado '] == 'EUA'].sort_values(
    by=['S&P Global ESG Score', 'Rendimiento'], ascending=False).head(5)
```
